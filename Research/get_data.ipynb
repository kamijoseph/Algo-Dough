## downloading data from yfinance

```python
import pandas as pd
import yfinance as yf
```

```python
ticker = "MSFT"
```

```python
# downloading the data
data = yf.download(
    ticker,
    start="2015-01-01",
    end="2026-08-27",
    interval="1d",
    auto_adjust=False,
    actions=True
)
```
