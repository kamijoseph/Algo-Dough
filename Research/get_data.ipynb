## downloading data from yfinance

```python
import pandas as pd
import yfinance as yf
```

```python
ticker = "MSFT"
```

```python
# downloading the data
data = yf.download(
    ticker,
    start="2015-01-01",
    end="2026-08-27",
    interval="1d",
    auto_adjust=False,
    actions=True
)
```

```python
data.head()
```

```python
data.info()
```

```python
data.head()
```

```python
# flattening
data.columns = data.columns.get_level_values(0)
```

```python
data.head()
```

```python
print(data.columns)
print(data.index.name)
print(data.dtypes)
```

```python
data.columns.name = None
```

```python
data.head()
```

```python
data = data.reset_index()
```

```python
data.head()
```

```python
# rearranging the data into ohlcv
import pandas as pd

# Remove column-axis name
data.columns.name = None

# Ensure Date is the index name
data.index = pd.to_datetime(data.index)
data.index.name = "Date"

# Put columns in conventional market-data order
data = data[
    [
        "Open",
        "High",
        "Low",
        "Close",
        "Adj Close",
        "Volume",
        "Dividends",
        "Stock Splits"
    ]
]

# Sort chronologically
data = data.sort_index()

# Optional: display prices to 2 decimal places
pd.options.display.float_format = "{:.2f}".format

print(data.head())
```

```python
data.to_csv("../Data/raw/MSFT.csv")
```
